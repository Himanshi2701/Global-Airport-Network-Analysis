# Airport Network Analysis — Data Exploration

This notebook introduces the OpenFlights airport and route datasets used in the network analysis.

## Objective

The objective of this notebook is to examine the structure and quality of the OpenFlights datasets before constructing the airport transportation network.

Specifically, this notebook:

- Introduces the airport and route datasets.
- Assigns descriptive column names.
- Evaluates data quality and missing values.
- Summarizes the geographical coverage of the airport dataset.
- Verifies that the data are suitable for network construction.

```python
from pathlib import Path
import pandas as pd

PROJECT_DIR = Path.cwd()
if not (PROJECT_DIR / 'Data').exists():
    PROJECT_DIR = PROJECT_DIR.parent
DATA_DIR = PROJECT_DIR / 'Data'
pd.set_option('display.max_columns', None)
print('Using data folder:', DATA_DIR)
```

## 1. Load the datasets

The files do not include headers, so descriptive column names are assigned after loading.

```python
airport_columns = [
    'airport_id', 'name', 'city', 'country', 'iata', 'icao', 'latitude',
    'longitude', 'altitude', 'timezone', 'dst', 'tz_database', 'type', 'source'
]
route_columns = [
    'airline', 'airline_id', 'source_iata', 'source_airport_id',
    'destination_iata', 'destination_airport_id', 'codeshare', 'stops', 'equipment'
]

airports = pd.read_csv(DATA_DIR / 'airports.dat', header=None, names=airport_columns)
routes = pd.read_csv(DATA_DIR / 'routes.dat', header=None, names=route_columns)

print(f'Airports: {airports.shape[0]:,} rows × {airports.shape[1]} columns')
print(f'Routes:   {routes.shape[0]:,} rows x {routes.shape[1]} columns')
```

## 2. Data quality checks

OpenFlights uses `\N` for missing values. Route records whose endpoint IDs are unknown or absent must be excluded before graph construction.

```python
print('Missing-value markers by airport column:')
display((airports == r'\N').sum().sort_values(ascending=False).to_frame('count'))
print('\nMissing-value markers by route column:')
display((routes == r'\N').sum().sort_values(ascending=False).to_frame('count'))
```

```python
top_countries = airports['country'].value_counts().head(10).rename_axis('country').reset_index(name='airports')
top_countries
```

## Observations

- The airport file contains airport attributes such as location, IATA code, and country.
- The routes file represents directed connections between source and destination airports.
- The next notebook cleans endpoint identifiers and builds the directed graph.

The exploratory analysis confirms that the OpenFlights datasets contain sufficient information to construct a global airport transportation network.

Although a small number of missing values are present in descriptive attributes, the datasets provide comprehensive airport identifiers, geographic coordinates, and route information required for network analysis.
